Data Visulization


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({
  "Name":["Rahul",  "Neha",  "Priya",  "karan" , "Riya"],
  "Age":[25 , 32 , 22 , 24 , 23],
  "Gender":['Male' , 'Female' , 'Female' , 'Male' , 'Female'],
  "Class":[1 , 2 , 1 , 1 , 3],
  "Fare":[500 , 800 , 1200 , 650 , 1800]
})

print(df)
print(df.dtypes)
```

```python
# Bar Chart: Total count for each Gender

gender_count = df["Gender"].value_counts()

plt.figure(facecolor="lightyellow")
axis = gender_count.plot(kind="bar", color=["skyblue", "lavender"])
axis.set_facecolor("white")
plt.title("Total Person by Gender")
plt.xlabel("Gender", color = "red")
plt.ylabel("Count", color = "green")
plt.show()
```
