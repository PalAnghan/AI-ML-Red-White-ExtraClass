```python
import pandas as pd 
import numpy as np 
```

```python
sales_data = pd.DataFrame({
  "Region":["South" , "South" , "North" , "North" , "East" , "East"],
  "Product":["Laptop",  "Mobile" , "Laptop" , "Mobile" , "Laptop" , "Mobile"],
  "Sales":[12000 , 10000 , 13000 , 11000 , 12500 , 10500]
})

print(sales_data)
```

```python
region_total = sales_data.groupby("Region")["Sales"].sum()

print(region_total)
```

```python
# Group students by class
# Find average marks per class

student_data = pd.DataFrame({
  "Name":["Rahul" , "Amit" , "Priya" , "Neha",  "Kiran",  "Riya"],
  "Class":["A" , "A" , "A" , "C" , "C" , "B"],
  "Marks":[80 , 90 , 75 , 88 , 88 , 92]
})

print(student_data)
```

```python
# findinig avaerage to we use mean..

class_average = student_data.groupby("Class")["Marks"].mean()

print(class_average)
```

```python
# Using transform()

# Create new column showing sales as percentage of total region sales

sales_data['Region_total'] = sales_data.groupby("Region")["Sales"].transform("sum")

print(sales_data)
```

```python
sales_data["Sales_Percentage"] = sales_data["Sales"] / sales_data["Region_total"] * 100

print(sales_data)
```

```python
# Group DataFrame by customer_id

# calculate each customer maximum order amount.

orders = pd.DataFrame({
  "customer_id":[101 , 101 , 101 , 102 , 102 , 103 , 103],
  "order_id":["0001",  "0002" , "0003" , "0004" , "0005" , "0006" , "0007"],
  "order_amount":[500 , 800 , 600 , 1100 , 700 , 400 , 900]
})

print(orders)
```

```python
customer_max = orders.groupby("customer_id")["order_amount"].max()

print(customer_max)
```
