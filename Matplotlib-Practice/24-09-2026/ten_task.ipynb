```python
import matplotlib.pyplot as plt
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
ac = [50, 60, 150, 200, 250, 220]
refrigerator = [40, 40, 45, 45, 50, 50]
lights = [30, 30, 25, 25, 20, 20]
other_appliances = [40, 45, 40, 50, 45, 50]

plt.figure(figsize=(10, 6))

plt.stackplot(months, ac, refrigerator, lights, other_appliances, 
              labels=['AC', 'Refrigerator', 'Lights', 'Other Appliances'],
              colors=['#1f77b4', '#aec7e8', '#ff7f0e', '#ffbb78'], 
              alpha=0.8)

plt.title('Monthly Electricity Consumption by Appliance Category', fontsize=14, fontweight='bold')
plt.xlabel('Months', fontsize=12)
plt.ylabel('Electricity Units (kWh)', fontsize=12)

plt.legend()
plt.show()
```
