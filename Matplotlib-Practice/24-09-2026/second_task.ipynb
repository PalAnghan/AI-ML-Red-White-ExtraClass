Task: Monthly Product Sales Analysis

Definition:
Create a Stackplot using Matplotlib to analyze how different product categories contribute to total monthly sales.

Real-time scenario:

You are working as a Sales Analyst. Your company sells Electronics, Clothing, and Grocery products. Management wants to see how each category contributes to overall sales throughout the year.

Perform these tasks:

1. Create data for 6 months.
2. Use 3 categories:
    * Electronics
    * Clothing
    * Grocery
3. Create a Stackplot.
4. Add:
    * Title: Monthly Product Sales
    * X-axis: Month
    * Y-axis: Sales
    * Legend
5. Display the chart.

```python
import matplotlib.pyplot as plt
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']

electronics = [15000, 18000, 16000, 20000, 22000, 25000]
clothing = [8000, 9500, 11000, 10500, 13000, 14000]
grocery = [12000, 12500, 13000, 13500, 14000, 15000]

plt.stackplot(
    months,
    electronics,
    clothing,
    grocery,
    labels=[
        "electronics",
        "clothing",
        "grocery"
    ],
    colors=[
        "blue",
        "pink",
        "green"
    ]
)
plt.title('Monthly Product Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.legend(loc='upper left')


plt.show()
```
