```python
import matplotlib.pyplot as plt
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
tutorials = [5000, 5200, 6000, 5800, 6500, 7000]
shorts = [10000, 15000, 18000, 25000, 30000, 35000]
reviews = [2000, 2500, 2200, 3000, 2800, 3500]

plt.figure(figsize=(10,6))

plt.stackplot(months, tutorials, shorts, reviews, 
              labels=['Tutorials', 'Shorts', 'Reviews'],
              colors=['#4daf4a', '#e41a1c', '#377eb8'], 
              alpha=0.8)

plt.title('Monthly YouTube Channel Views by Content Type', fontsize=14, fontweight='bold')
plt.xlabel('Months', fontsize=12)
plt.ylabel('Number of Views', fontsize=12)

plt.legend()
plt.show()
```
