Create a Stackplot using Matplotlib to visualize how different traffic sources contribute to total website visitors over several months.

Perform these tasks:

1. Create data for 6 months.
2. Use 3 traffic sources:
    * Google Search
    * Social Media
    * Direct Visitors
3. Create a stackplot with months on the X-axis and visitors on the Y-axis.
4. Add:
    * Chart title
    * X-axis label: Month
    * Y-axis label: Visitors
    * Legend
5. Display the chart.

```python
import matplotlib.pyplot as plt
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']

google_search = [1000, 1200, 1400, 1300, 1600, 1800]
social_media = [400, 500, 700, 800, 1000, 1100]
direct_visitors = [300, 350, 400, 450, 500, 600]

plt.figure(figsize=(10, 5))

plt.stackplot(
    months,
    google_search,
    social_media,
    direct_visitors,
    labels=[
        "google_search",
        "social_media",
        "direct_visitors"
    ],
    colors=[
        "lightblue",
        "lightgreen",
        "lightpink"
    ]
)

plt.title("Monthly Website Traffic Breakdown by Source")
plt.xlabel("Month")
plt.ylabel("Vistitors")

plt.legend()
plt.show()
```
