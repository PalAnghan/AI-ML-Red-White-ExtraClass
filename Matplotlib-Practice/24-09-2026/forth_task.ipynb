```python
import matplotlib.pyplot as plt
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
videos = [20, 25, 22, 28, 30, 35]
games = [15, 18, 20, 25, 22, 20]
chat = [10, 12, 15, 14, 18, 20]
shopping = [5, 8, 10, 12, 15, 18]

plt.stackplot(
    months,
    videos,
    games,
    chat,
    shopping,
    labels=[
        "videos",
        "games",
        "chat",
        "shopping"
    ],
    colors=['#e41a1c', '#377eb8', '#4daf4a', '#984ea3'],
    alpha = 0.7
)

plt.title('Monthly Mobile App Usage Breakdown', fontsize=14, fontweight='bold')
plt.xlabel('Months', fontsize=12)
plt.ylabel('Total Usage Hours', fontsize=12)

plt.legend(loc="upper left")



plt.show()
```
