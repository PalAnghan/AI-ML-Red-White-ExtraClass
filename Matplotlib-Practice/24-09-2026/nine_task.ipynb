```python
import matplotlib.pyplot as plt
```

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
walking = [30, 45, 30, 40, 50, 60, 45]
running = [15, 0, 20, 15, 0, 30, 20]
cycling = [0, 30, 0, 0, 45, 60, 0]


plt.figure(figsize=(10, 6))

plt.stackplot(
    days,
    walking,
    running,
    cycling,
    labels=[
        "walking",
        "running",
        "cycling"
    ],
    colors=['#66c2a5', '#fc8d62', '#8da0cb'],
    alpha = 0.8
    )
plt.title('Daily Fitness Activity Duration', fontsize=14, fontweight='bold')
plt.xlabel('Days of the Week', fontsize=12)
plt.ylabel('Activity Duration (Minutes)', fontsize=12)

plt.legend()
plt.show()
```
