```python
import matplotlib.pyplot as plt
```

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
python = [2, 1, 3, 2, 2, 4, 1]
sql = [1, 2, 1, 1, 2, 1, 1]
mathematics = [1, 1, 2, 1, 1, 2, 3]
revision = [0, 1, 0, 1, 1, 2, 2]

plt.figure(figsize=(10, 6))

plt.stackplot(
    days,
    python,
    sql,
    mathematics,
    revision,
    labels=[
        "python",
        "sql",
        "mathematics",
        "revision"
    ],
    colors=['#2ca02c', '#1f77b4', '#ff7f0e', '#d62728'],
)

plt.title('Weekly Study Time Distribution', fontsize=14, fontweight='bold')
plt.xlabel('Days of the Week', fontsize=12)
plt.ylabel('Study Hours', fontsize=12)



plt.legend(loc='upper left')


plt.tight_layout()
plt.show()
```
