```python
import matplotlib.pyplot as plt
```

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
breakfast = [300, 320, 310, 330, 350, 400, 420]
lunch = [600, 580, 620, 650, 700, 850, 900]
dinner = [800, 750, 820, 850, 1000, 1200, 1150]

plt.figure(figsize=(10, 6))
plt.stackplot(days, breakfast, lunch, dinner, 
              labels=['Breakfast', 'Lunch', 'Dinner'],
              colors=['#ffcc5c', '#96ceb4', '#ff6f69'], 
              alpha=0.8)

plt.title('Daily Restaurant Sales by Meal Type', fontsize=14, fontweight='bold')
plt.xlabel('Days of the Week', fontsize=12)
plt.ylabel('Sales Amount ($)', fontsize=12)

plt.legend()
plt.show()
```
