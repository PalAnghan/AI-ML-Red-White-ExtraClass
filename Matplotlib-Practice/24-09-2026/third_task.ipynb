```python
import matplotlib.pyplot as plt
```

```python
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']
coding = [5, 4, 6, 5, 4]
meetings = [2, 3, 1, 2, 2]
training = [1, 1, 1, 1, 2]

plt.stackplot(
    days,
    coding,
    meetings,
    training,
    labels=[
        "coding",
        "meetings",
        "training"
    ],
    colors=[
        "lightblue",
        "pink",
        "green"
    ]
)

plt.title('Employee Weekly Work Hours Distribution', fontsize=14, fontweight='bold')

plt.xlabel("Days of the week")
plt.ylabel("Hours")

plt.legend()
plt.shows()
```
