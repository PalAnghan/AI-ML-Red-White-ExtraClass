```python
import matplotlib.pyplot as plt
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
electronics = [500, 520, 600, 580, 650, 700]
fashion = [800, 850, 820, 900, 950, 1000]
grocery = [1200, 1150, 1250, 1300, 1280, 1350]

plt.stackplot(months, electronics, fashion, grocery, 
              labels=['Electronics', 'Fashion', 'Grocery'],
              colors=['#1f77b4', '#ff7f0e', '#2ca02c'], 
              alpha=0.8)
plt.title('Monthly E-Commerce Order Volume by Category', fontsize=14, fontweight='bold')
plt.xlabel('Months', fontsize=12)
plt.ylabel('Number of Orders', fontsize=12)

plt.legend()

plt.show()
```
