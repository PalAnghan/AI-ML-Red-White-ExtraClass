```python
# Bar chart

import matplotlib.pyplot as plt
```

```python
product = ['A', 'B', 'C','D']
sales = [1000, 1500, 800, 1200]

plt.bar(product, sales, color= "lightblue", label='Sales 2026')
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('Product Sale Comparison ')
plt.legend()
plt.show()
```

```python
# pie chart

regions = ['North', 'South', 'East', 'West']
revenue = [3000, 2000, 1500, 1000]

plt.pie(revenue,labels = regions, autopct='%1.1f%%' , colors = ['gold', 'skyblue', 'lightgreen', 'coral'])
plt.title('Revenue Contribution By Region')
plt.show()
```

```python
# Histogram 

scores = [23, 34, 36, 23, 56 , 34, 23, 78, 123, 23, 67 , 45, 56, 78, 23, 6, 12, 67, 34, 78 ,98]

plt.hist(scores, bins=5, color='pink', edgecolor='black')

plt.xlabel('Score Range')
plt.ylabel('Number of Students')

plt.title('Score Distribution of Students')
plt.show()
```
