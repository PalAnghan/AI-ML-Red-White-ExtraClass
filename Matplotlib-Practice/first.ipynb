Introduction to MATPLOTLIB

Matplotlib is a python library used to create charts and visualizations.

Comman charts:

1. Bar Chart

2. Line Chart

3. Scatter Chart

4. Pie Chart

5. Histogram 




```python
#Common keyword and Functions: 

# plt.figure() : Create a new figure.
# plt.title() : Add chart title.
# plt.xlabel() : Add X-axis label.
# plt.ylabel() : Add Y-axis label.
# plt.xticks() : Change X-axis tick labels.
# plt.yticks() : Change Y-axis tick labels.
# plt.legend() : Display legend.
# plt.grid() : Display grid lines.
# plt.show() : Display the chart.
# plt.savefig() : Save chart as an image.
# plt.xlim() : Get X-axis limit
# plt.ylim() : Get Y-axis limit
# plt.text() : Add text to chart.
# plt.annotate() : Add annotation to chart.
# aplha : Controls transparency.
# color : Set color.
# label : Set legend label.
# marker : Set point style.
# linestyle : set line style.
# linewidth : Set line thickness.
# figsize : Set chart width and height.
```

```python
import matplotlib.pyplot as plt
```

```python
# Bar Chart

categories = ["Rent", "Groceries", "Utilities", "Entertainment", "Savings"]
amount = [1200, 450, 180, 250, 500]

plt.figure(figsize=(8 ,5))
plt.bar(categories , amount , color="teal", label= "Expenses ($)")
plt.title("Monthly Household Expenses")
plt.xlabel("Categories")
plt.ylabel("Amount Spent ($)")
plt.legend()
plt.show()
```

```python
# Line Chart

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [22, 24, 21, 25, 28, 27, 23]

plt.figure(figsize= (10 , 5))

plt.plot(
    days,
    temperature,
    marker="o",
    linestyle="-",
    linewidth=2,
    label = "Temperature (°C)"
)

plt.title("Weekly Temperature Variation")
plt.xlabel("Day of the Week")
plt.ylabel("Temperature (°C)")

plt.grid(True)
plt.legend()

plt.show()

```

```python
# SCATTER CART

study_hours = [1.5, 2.0, 2.5, 3.2, 3.8, 4.0, 4.5, 5.1, 5.5, 6.0, 6.5, 7.0]
test_scores = [52, 55, 60, 65, 70, 72, 78, 83, 85, 88, 92, 95]

plt.figure(figsize=(8, 5))

plt.scatter(
    study_hours,
    test_scores,
    color="green",
    marker="o",
    label="Student Performance"
)

# Add titles and formatting
plt.title("Study Hours vs. Test Scores")
plt.xlabel("Hours Spent Studying")
plt.ylabel("Test Score (%)")
plt.grid(True)
plt.legend()

# Display chart
plt.show()
```

```python
# Pie Chart
activities = ["Work", "Sleep", "Exercise", "Entertainment", "Other"]
hours = [8, 7, 1.5, 4.5, 3]

# Create figure
plt.figure(figsize=(7, 7))

# Create pie chart
plt.pie(
    hours,
    labels=activities,
    autopct="%1.1f%%",  # Displays percentage values on each slice
    startangle=90,
    colors=["#4CAF50", "#2196F3", "#FF9800", "#9C27B0", "#607D8B"]
)

# Add title
plt.title("Daily Time Distribution")

# Display chart
plt.show()
```

```python
exam_scores = [
    55, 62, 67, 70, 71, 73, 74, 75, 76, 78, 79, 80, 81, 82, 83, 84,
    85, 86, 87, 88, 89, 90, 91, 92, 93, 95, 96, 98, 100
]

# Create figure
plt.figure(figsize=(8, 5))

# Create histogram
plt.hist(
    exam_scores,
    bins=8,
    color="purple",
    edgecolor="black",
    alpha=0.7,
    linewidth=1.2
)

# Add titles and labels
plt.title("Exam Score Distribution")
plt.xlabel("Scores")
plt.ylabel("Number of Students")
plt.grid(axis="y")

# Display chart
plt.show()
```
