```python
import pandas as pd
```

```python

missing_df = pd.DataFrame({
  "name":["Rahul" , "Ronak" , "Rajiv" , "Raj" , "Rakesh"],
  "age":[20 , None , 30 , 35 , 40],
  "salary":[30000 , 45000 , 40000 , 50000 , 25000],
  "city":["Surat",  "Rajkot" , "Vadodara" , None , "Ahmedabad"]
})

print(missing_df)
```

```python
# isnull() = check missing values

print(missing_df.isnull())
```

```python
#isna() = Same purpose as isnull()

print(missing_df.isna()) 
```

```python
#fillna() = Fill missing value

filled_df = missing_df.copy()

filled_df['age'] = filled_df['age'].fillna(100)
print(filled_df)

missing_df.loc[1 , "age"] = 150

print(missing_df)
```

```python
#dropna = remove row missing value

drop_df = missing_df.dropna()
print(drop_df)
```

```python
duplicate_df = pd.DataFrame({
  "name":["Rahul" , "Ronak" , "Rahul" , "Raj" , "Rakesh"],
  "age":[20 , None , 20 , 35 , 40],
  "salary":[20 , 45000 , 20 , 50000 , 25000],
  "city":["Rahul",  "Rajkot" , "Rahul" , None , "Surat"]
})

print(duplicate_df)
```

```python
# duplicated() = check duplicate rows

print(duplicate_df.duplicated()) # return True and False

print(duplicate_df.drop_duplicates())
```


CHECK AND CHANGE DATA TYPES

```python
data_types_df = pd.DataFrame({
  'name':['A' , 'B' , 'C' , 'D'],
  'age':[20 , 25 , 21 , 30],
  'salary':[10000 , 21000 , 40000 , 50000]
})

print(data_types_df)

print(data_types_df.dtypes) # find for DataTypes
```

```python
data_types_df['age'] = data_types_df['age'].astype(float) # data type change into float

print(data_types_df.dtypes)
```
