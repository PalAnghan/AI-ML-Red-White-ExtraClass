```python
import pandas as pd
```

```python
# Employee working hours by department and project

working_hours  = pd.DataFrame({
  "Department":["IT",  "IT" , "HR" , "HR" , "Sales" , "Sales"],
  "Project":["Website" , "Mobile App",  "Recruitment" , "Training" , "Marketing" , "CRM"],
  "Hours":[120 , 120 , 80 , 90 , 90 , 110]
})

print(working_hours)
```

```python
working_pivoat = working_hours.pivot_table(
    index="Project",
    columns="Department",
    values="Hours", 
    aggfunc="sum",
    fill_value=0
)

print(working_pivoat)
```

```python
# Average prices of diffrent fruits types by region

fruits = pd.DataFrame({
  "Region":["South" , "South" , "North" , "North" , "East" , "East"],
  "Fruit":["Coconut",  "Banana" , "Apple" , "Graps" , "Orange" , "Mango"],
  "Price":[120 , 100 , 130 , 110 , 125 , 105]
})

print(fruits)
```

```python
fruits_pivot = fruits.pivot_table(
    index="Region",
    columns="Fruit",
    values="Price",
    aggfunc=sum,
    fill_value=0
)

print(fruits_pivot)
```

```python

# Re-indexing & Altering Lables

# Re-index employee IDs to start from 1001

employee = pd.DataFrame(
  {
    "Employee_ID":[1 , 2 , 3 , 4 , 5],
    "Name":["Rahul" , "Pal" , "Preet" , "Popatlal" , "Prachi"],
    "Salary":[30000 , 50000 , 40000 , 32000 , 450000]
  }
)

print(employee)
```

```python
employee.index = range(10001, 10006)

employee["Employee_ID"] = employee.index

print(employee)
```

```python
# new index

new_index = range(10001, 10006)

employee_reindex = employee.reindex(new_index)

print(employee_reindex)
```

```python
# Rename column labels
employees = employee.rename(columns={
  "Employee_ID" : "Emp_ID",
  "Name" : "Employee_Name",
  "Salary" : "Employee_Salary"
})

print(employees)
```

```python
# Change index of DataFrame contining customer names

# to uppercase

customer = pd.DataFrame({
  "City" :["Surat" , "Bhopal" , "Godhara" , "Rajkot",  "Vadodara"],
  "Purchase":[2000 , 5000 , 7000 , 6000 , 4500]
})

customer.index = ["rahul", "preet" , "pal", "prachi", "amit"]
```

```python
# uopper case

customer["City"] = customer["City"].str.upper()
print(customer)
```

```python
# lower case

customer["City"] = customer["City"].str.lower()
print(customer)
```

```python
# captilice

customer['City'] = customer['City'].str.capitalize()
print(customer)

print("===================================")

customer['City'] = customer['City'].str.replace("bho", "",  regex=False)
print(customer)
```
