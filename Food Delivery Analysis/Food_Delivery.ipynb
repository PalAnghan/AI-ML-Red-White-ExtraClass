1 : Data Preparation 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("food_delivery_data.csv")

df.head()
    
```

```python
#Basic Statistics

# Total number of orders
print("Total Orders:", df["Order_ID"].count())

# Total amount
print("Total Amount:", df["Price_INR"].sum())

# Highest amount
print("Highest Amount:", df["Price_INR"].max())

# Lowest amount
print("Lowest Amount:", df["Price_INR"].min())

# Average / Mean amount
print("Mean Amount:", df["Price_INR"].mean())

# Median amount
print("Median Amount:", df["Price_INR"].median())

# Standard deviation
print("Standard Deviation:", df["Price_INR"].std())

# Variance
print("Variance:", df["Price_INR"].var())

# Number of unique prices
print("Unique Prices:", df["Price_INR"].nunique())

# Most common price
print("Most Common Price:", df["Price_INR"].mode()[0])
```

```python
#Quantity Analysis

print("Total Quantity:", df["Quantity"].sum())

print("Highest Quantity:", df["Quantity"].max())

print("Lowest Quantity:", df["Quantity"].min())

print("Average Quantity:", df["Quantity"].mean())

print("Median Quantity:", df["Quantity"].median())
```

```python
#Rating Analysis

print("Highest Rating:", df["Rating"].max())

print("Lowest Rating:", df["Rating"].min())

print("Average Rating:", df["Rating"].mean())

print("Median Rating:", df["Rating"].median())

print("Most Common Rating:", df["Rating"].mode()[0])
```

```python
#Category / City Counts

# Orders by city
print(df["City"].value_counts())

# Orders by restaurant
print(df["Restaurant"].value_counts())

# Orders by food item
print(df["Food_Item"].value_counts())

# Orders by category
print(df["Category"].value_counts())

# Orders by payment method
print(df["Payment_Method"].value_counts())

# Orders by status
print(df["Order_Status"].value_counts())
```

```python
# Box plot

plt.figure(figsize=(18 , 6))

sns.boxplot(
    x='Food_Item',
    y='Price_INR',
    data = df,
    hue="Food_Item",
    orient= 'x'
)

plt.title("Total Food Item and Price of Food")

plt.xlabel("Food_Item")
plt.ylabel("Prices of Food")


plt.show()
```

```python
# Heatmap

numeric_columns = df.select_dtypes(include=np.number)

correlation = numeric_columns.corr()

plt.figure(figsize=(6, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.xticks(rotation=2)
plt.yticks(rotation=45)

plt.title("Correlation Between Numeric Columns")

plt.show()
```
